```python
import pandas as pd

from datasets import load_dataset
```

```python
block_ds = load_dataset(
  "RevolutionCrossroads/RevX_OCR_results",
  split="train",
  name="block_level",
  streaming=True
)
```

```python
for row in block_ds.take(1):
  display(row)
```

```python
def by_news(row):
  return row["collection"] == "chronicling_america"

def by_text(row):
  return row["block_type"] == "Text"

def by_news_text(row):
  return row["collection"] == "chronicling_america" and row["block_type"] == "Text"

block_news_text_ds = block_ds.filter(by_news_text)
```

```python
block_news_text = list(block_news_text_ds.take(500))

len(block_news_text)
```

```python
import json

ofp = open("./chronicling_america_text_500.json", "w")
json.dump(block_news_text, ofp)
ofp.close()
```
